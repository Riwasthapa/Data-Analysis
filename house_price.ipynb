```python
X = df.drop('TARGET(PRICE_IN_LACS)', axis = 1)
y = df['TARGET(PRICE_IN_LACS)']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size = 0.8, random_state = 42)
```

```python
len(X_train), len(X_test)
```

```python
X_train.head()
```

```python
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()
```

```python
from sklearn.preprocessing import StandardScaler

scaler_x = StandardScaler()
scaler_y = StandardScaler()

X_train_scaled[['BHK_NO.', 'SQUARE_FT']] = scaler_x.fit_transform(X_train_scaled[['BHK_NO.', 'SQUARE_FT']])
X_test_scaled[['BHK_NO.', 'SQUARE_FT']] = scaler_x.transform(X_test_scaled[['BHK_NO.', 'SQUARE_FT']])
y_train_scaled = scaler_y.fit_transform(y_train.values.reshape(-1,1))
y_test_scaled = scaler_y.transform(y_test.values.reshape(-1,1))
```

```python
X_train_scaled[['BHK_NO.', 'SQUARE_FT']]
```

```python
scaler_x.fit_transform(X_train_scaled[['BHK_NO.', 'SQUARE_FT']])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

model = LinearRegression()
model.fit(X_train, y_train)

# Train Prediction
train_pred = model.predict(X_train)
train_mse = mean_squared_error(y_train, train_pred)
train_rmse = np.sqrt(train_mse)
train_score = r2_score(y_train, train_pred)


# Test Prediction
test_pred = model.predict(X_test)
test_mse = mean_squared_error(y_test, test_pred)
test_rmse = np.sqrt(test_mse)
test_score = r2_score(y_test, test_pred)


train_rmse, train_score, test_rmse, test_score
```

Linear Regression wih Feature Scaling

```python
model.fit(X_train_scaled, y_train_scaled)

# Train Prediction
train_pred = model.predict(X_train_scaled)
train_pred = scaler_y.inverse_transform(train_pred)
train_mse = mean_squared_error(y_train, train_pred)
train_rmse = np.sqrt(train_mse)
train_score = r2_score(y_train, train_pred)


# Test Prediction
test_pred = model.predict(X_test_scaled)
test_pred = scaler_y.inverse_transform(test_pred)
test_mse = mean_squared_error(y_test, test_pred)
test_rmse = np.sqrt(test_mse)
test_score = r2_score(y_test, test_pred)


train_rmse, train_score, test_rmse, test_score
```

DT without Feature scaling

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(random_state = 42)
model.fit(X_train, y_train)

# Train Prediction
train_pred = model.predict(X_train)
train_mse = mean_squared_error(y_train, train_pred)
train_rmse = np.sqrt(train_mse)
train_score = r2_score(y_train, train_pred)


# Test Prediction
test_pred = model.predict(X_test)
test_mse = mean_squared_error(y_test, test_pred)
test_rmse = np.sqrt(test_mse)
test_score = r2_score(y_test, test_pred)


train_rmse, train_score, test_rmse, test_score
```


DT with Feature scaling

```python
model.fit(X_train_scaled, y_train_scaled)

# Train Prediction
train_pred = model.predict(X_train_scaled)
train_pred = scaler_y.inverse_transform(train_pred.reshape(-1,1))
train_mse = mean_squared_error(y_train, train_pred)
train_rmse = np.sqrt(train_mse)
train_score = r2_score(y_train, train_pred)


# Test Prediction
test_pred = model.predict(X_test_scaled)
test_pred = scaler_y.inverse_transform(test_pred.reshape(-1,1))
test_mse = mean_squared_error(y_test, test_pred)
test_rmse = np.sqrt(test_mse)
test_score = r2_score(y_test, test_pred)


train_rmse, train_score, test_rmse, test_score
```

GBoost

```python
from sklearn.ensemble import GradientBoostingRegressor

gboost = GradientBoostingRegressor(random_state = 42)
gboost.fit(X_train, y_train)

# Train Prediction
train_pred = gboost.predict(X_train)
train_mse = mean_squared_error(y_train, train_pred)
train_rmse = np.sqrt(train_mse)
train_score = r2_score(y_train, train_pred)


# Test Prediction
test_pred = gboost.predict(X_test)
test_mse = mean_squared_error(y_test, test_pred)
test_rmse = np.sqrt(test_mse)
test_score = r2_score(y_test, test_pred)


train_rmse, train_score, test_rmse, test_score
```
